```python
import json
import mcstas2json as m2j

result = m2j.mcstas_to_sample_relative_json(
    "ESS_MIRACLES_kafkatubes.instr",
    sample_component="Sample",
    exclude_components=[
        "ISCS",'chopper_settings','origin','VS']
)

with open("sample_relative_components.json", "w") as f:
    json.dump(result, f, indent=2)
```

```python
#first part of detectors (9.4 deg->85.6 deg)
import numpy as np
import matplotlib.pyplot as plt
xdelta=3.24
x0=9.48+xdelta/2# addressing center of tubes
rad=0.23
n=[]
x=[]
y=[]
z=[]
for hi in range(150):
    for hia in range(24):
        n.append(len(n))
        z.append((hi/10-7.45)/100)
        angle=x0+hia*xdelta
        x.append(float(np.sin(angle/180*np.pi)*rad))
        y.append(float(np.cos(angle/180*np.pi)*rad))
plt.figure()
plt.plot(x,y)
print(n)
print(x)
print(y)
print(z)
```
